```python
# ==============================================================================
# ML PRACTICAL 4: GRADIENT DESCENT ALGORITHM
# Objective: Find local minima of y = (x + 3)^2 starting at x = 2 using Gradient Descent
# ==============================================================================

# 1. Import Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Define Objective Function and its First Derivative
# Objective function: y = (x + 3)^2
def func(x):
    return (x + 3) ** 2

# First derivative: dy/dx = 2 * (x + 3)
def derivative(x):
    return 2 * (x + 3)

# 3. Gradient Descent Implementation
def gradient_descent(starting_x=2.0, learning_rate=0.1, num_iterations=50):
    x = starting_x
    history = []
    
    # Record initial position (Iteration 0)
    history.append({'Iteration': 0, 'x': x, 'f(x)': func(x), 'Gradient': derivative(x)})
    
    for i in range(1, num_iterations + 1):
        grad = derivative(x)
        x = x - learning_rate * grad
        history.append({'Iteration': i, 'x': x, 'f(x)': func(x), 'Gradient': grad})
        
    return x, pd.DataFrame(history)

# Execute Gradient Descent
starting_x = 2.0
learning_rate = 0.1
num_iterations = 50

min_x, history_df = gradient_descent(starting_x, learning_rate, num_iterations)

print(f"Starting Point: x = {starting_x}, f(x) = {func(starting_x)}")
print(f"Local Minimum occurs at: x = {min_x:.6f}")
print(f"Minimum Function Value: f(x) = {func(min_x):.6e}")

# 4. Display Milestone Iterations
milestones = [0, 1, 2, 3, 5, 10, 20, 30, 40, 50]
print("\n--- Milestone Iterations ---")
print(history_df.loc[milestones].reset_index(drop=True))

# 5. Visualizations
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Function Curve and Optimization Trajectory
x_curve = np.linspace(-6, 3, 300)
y_curve = func(x_curve)

axes[0].plot(x_curve, y_curve, label=r'$f(x) = (x + 3)^2$', color='#2980b9', lw=2.5)
axes[0].plot(history_df['x'], history_df['f(x)'], 'o--', color='#e74c3c', label='Gradient Descent Steps', markersize=4, alpha=0.8)
axes[0].plot(starting_x, func(starting_x), 's', color='#27ae60', markersize=9, label=f'Start: ({starting_x}, {func(starting_x)})')
axes[0].plot(min_x, func(min_x), '*', color='#f39c12', markersize=14, label=f'Min: ({min_x:.2f}, {func(min_x):.2f})')

axes[0].set_title('Gradient Descent Trajectory on $y = (x+3)^2$', fontweight='bold')
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.6)

# Plot 2: Cost Function Reduction over Iterations
axes[1].plot(history_df['Iteration'], history_df['f(x)'], color='#8e44ad', lw=2.5)
axes[1].set_title('Cost Reduction: $f(x)$ vs. Iterations', fontweight='bold')
axes[1].set_xlabel('Iteration')
axes[1].set_ylabel('f(x) [Loss]')
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()

# ==============================================================================
# VIVA QUICK REFERENCE:
# 1. Update Rule: x_new = x_old - learning_rate * (dy/dx)
# 2. Learning Rate effect:
#    - Too large: Overshooting / divergence.
#    - Too small: Very slow convergence.
# 3. Stopping Criteria: Maximum iterations or gradient norm below tolerance (|grad| < eps).
# ==============================================================================
```
